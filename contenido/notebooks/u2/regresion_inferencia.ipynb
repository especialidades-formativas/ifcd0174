## 1. Regresión lineal simple: consumo según la carga

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
from scipy import stats

datos = pd.DataFrame({
    'consumo': [5.8, 7.5, 8.3, 9.7, 10.6, 12.2, 13.5, 15.7, 8.3, 10.1, 12.5, 14.4],  # kWh
    'carga':   [20, 30, 40, 50, 60, 70, 80, 90, 35, 55, 75, 85],                    # %
    'temp':    [18, 22, 20, 25, 19, 27, 24, 30, 28, 21, 17, 26],                    # ºC
})
simple = smf.ols('consumo ~ carga', data=datos).fit()
print(simple.params.round(3)); print('R2 =', round(simple.rsquared, 3))
```

```python
import matplotlib.pyplot as plt
fig, (a, b) = plt.subplots(1, 2, figsize=(9, 3.2))
a.scatter(datos.carga, datos.consumo, color='#1F3A40')
x = np.linspace(15, 95, 50); a.plot(x, simple.params.Intercept + simple.params.carga * x, color='#1C8C96')
a.set_xlabel('Carga (%)'); a.set_ylabel('Consumo (kWh)')
b.scatter(simple.fittedvalues, simple.resid, color='#1F3A40'); b.axhline(0, color='grey', ls='--')
b.set_xlabel('Valor ajustado'); b.set_ylabel('Residuo'); b.set_title('Residuos: deben repartirse al azar', fontsize=10)
plt.tight_layout(); plt.show()
```

## 2. Regresión múltiple: la salida completa (tabla 2.4)

```python
multiple = smf.ols('consumo ~ carga + temp', data=datos).fit()
print(multiple.summary())
```

```python
nuevo = pd.DataFrame({'carga': [65], 'temp': [22]})
print('Predicción:', round(multiple.predict(nuevo)[0], 2), 'kWh')
```

## 3. Intervalo de confianza y contraste (ejemplo 2.6)

```python
diam = np.array([50.03, 49.98, 50.05, 50.02, 50.04, 49.99, 50.06, 50.01, 50.03, 50.04])  # mm
n = len(diam); media = diam.mean(); et = diam.std(ddof=1) / np.sqrt(n)
t_crit = stats.t.ppf(0.975, n - 1)
print(f'IC 95 %: [{media - t_crit * et:.3f}, {media + t_crit * et:.3f}] mm')
print(stats.ttest_1samp(diam, 50.00))
```

¿Es relevante un desplazamiento de 0,025 mm? Depende de la tolerancia, no del valor *p* (Wasserstein y Lazar, 2016).

## 4. ANOVA de un factor y comparaciones de Tukey (ejemplo 2.5)

```python
A = [52, 55, 53, 56]; B = [58, 60, 57, 61]; C = [54, 53, 56, 55]   # piezas/hora
print(stats.f_oneway(A, B, C))
print(stats.tukey_hsd(A, B, C))
```

En la salida de Tukey, el grupo 0 es A, el 1 es B y el 2 es C.

> Seabold, S. y Perktold, J. (2010). Statsmodels: Econometric and Statistical Modeling with Python. *Proc. 9th Python in Science Conf.*, 92–96. https://doi.org/10.25080/Majora-92bf1922-011

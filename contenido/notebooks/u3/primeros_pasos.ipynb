```python
tolerancia = 0.04
medidas = [50.03, 49.95, 50.01, 50.06]
for m in medidas:
    if abs(m - 50.00) <= tolerancia:
        print(m, "conforme")
    else:
        print(m, "fuera de tolerancia")
```

## 3. Funciones

```python
def fuera_de_tolerancia(medidas, nominal, tolerancia=0.04):
    """Devuelve las medidas que se salen de nominal ± tolerancia."""
    return [m for m in medidas if abs(m - nominal) > tolerancia]

print(fuera_de_tolerancia(medidas, 50.00))
print(fuera_de_tolerancia(medidas, 50.00, tolerancia=0.05))
```

## 4. Comprensiones, lambdas y generadores

```python
print([x**2 for x in range(5)])

rendimiento = {"A": 54.0, "B": 59.0, "C": 54.5}
print({k: v for k, v in rendimiento.items() if v > 54})
print(max(rendimiento, key=lambda k: rendimiento[k]))
print(sorted(rendimiento, key=lambda k: rendimiento[k], reverse=True))
```

```python
def lotes(datos, tam):
    """Entrega los datos en bloques de tamaño tam."""
    for i in range(0, len(datos), tam):
        yield datos[i:i + tam]

print(list(lotes(list(range(7)), 3)))
```

## 5. Clases y excepciones

```python
class MediaPredictor:
    """Modelo de referencia: predice siempre la media de y."""

    def fit(self, X, y):
        self.media_ = sum(y) / len(y)
        return self

    def predict(self, X):
        return [self.media_ for _ in X]

modelo = MediaPredictor().fit([[1], [2], [3]], [10, 20, 30])
print(modelo.media_, modelo.predict([[4], [5]]))
```

```python
def leer_medida(texto):
    try:
        return float(texto.replace(",", "."))
    except ValueError:
        print(f"Valor no válido: {texto!r}")
        return None

print(leer_medida("50,02"))
print(leer_medida("n/d"))
```

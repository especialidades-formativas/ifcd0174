# U1-02 · Calcula la energía de un entrenamiento

*Machine Learning e Inteligencia Artificial (IFCD0174) · Unidad 1, apartado «Monitorizar el consumo»*

Patterson et al. (2022) proponen estimar la energía de un entrenamiento a partir de magnitudes fáciles de registrar:

$$E\,[\mathrm{kWh}] = t \cdot n \cdot \bar{P} \cdot \mathrm{PUE}$$

- $t$: horas de entrenamiento
- $n$: número de procesadores
- $\bar{P}$: potencia media de cada procesador, en kW
- PUE: energía total del centro de datos / energía de los equipos informáticos

Multiplicando $E$ por la intensidad de carbono de la red eléctrica se obtienen las emisiones.

> Patterson, D. et al. (2022). The Carbon Footprint of Machine Learning Training Will Plateau, Then Shrink. *Computer*, 55(7), 18–28. https://doi.org/10.1109/MC.2022.3148714

Ejecuta cada celda con **Mayús + Intro**. Puedes cambiar los valores y volver a ejecutar.

## 1. La fórmula como función de Python

```python
def energia_kwh(horas, procesadores, potencia_kw, pue):
    """Energía de un entrenamiento en kWh (Patterson et al., 2022)."""
    return horas * procesadores * potencia_kw * pue

def emisiones_kg(energia, intensidad_g_por_kwh):
    """Emisiones en kg de CO2 equivalente a partir de la intensidad de la red (g/kWh)."""
    return energia * intensidad_g_por_kwh / 1000
```

## 2. El ejemplo 1.2 del libro

Datos hipotéticos: 24 horas, 8 procesadores gráficos de 0,3 kW y un PUE de 1,5.

```python
E = energia_kwh(horas=24, procesadores=8, potencia_kw=0.3, pue=1.5)
print(f"Energía: {E:.1f} kWh")
```

## 3. Prueba con tus propios datos

La intensidad de carbono depende de la red y del momento: consúltala en la web de tu operador del sistema eléctrico. El valor siguiente es solo un ejemplo.

```python
horas = 24
procesadores = 8
potencia_kw = 0.3
pue = 1.5
intensidad = 200   # g CO2 eq por kWh (valor de ejemplo, cámbialo)

E = energia_kwh(horas, procesadores, potencia_kw, pue)
print(f"Energía:   {E:8.1f} kWh")
print(f"Emisiones: {emisiones_kg(E, intensidad):8.1f} kg CO2 eq")
```

## 4. ¿Cuánto importa el PUE?

Mismo entrenamiento en centros de datos más o menos eficientes.

```python
import matplotlib.pyplot as plt

pues = [1.1, 1.2, 1.4, 1.6, 1.8, 2.0]
energias = [energia_kwh(24, 8, 0.3, p) for p in pues]

plt.figure(figsize=(6, 3.5))
plt.bar([str(p) for p in pues], energias, color="#1C8C96")
plt.xlabel("PUE del centro de datos")
plt.ylabel("Energía (kWh)")
plt.title("Mismo entrenamiento, distinto PUE")
plt.tight_layout()
plt.show()
```
